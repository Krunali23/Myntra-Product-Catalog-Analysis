```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

df['Selling_Price'].fillna(df['Selling_Price'].median(), inplace=True)
```

```python
df['Selling_Price'].isnull().sum()
```

```python
plt.boxplot(df["Selling_Price"])
plt.show()
```

```python
Q1 = df['Selling_Price'].quantile(0.25)
Q3 = df['Selling_Price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Limit:", lower)
print("Upper Limit:", upper)
```

```python
df['Selling_Price'] = np.where(
    df['Selling_Price'] > upper,
    upper,
    df['Selling_Price']
)
```

## VISUALISATION

### `UNIVARIATE ANALYSIS`

### Category Analysis

```python
plt.figure(figsize = (6,4))
sns.countplot(data = df, x = 'Category',
              order = df['Category'].value_counts().index,palette = 'Set1')
plt.title('Distribution of products by Category')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation = 45)
plt.show()
```

 ## `INSIGHT`: 
- The count plot shows the distribution of products across different categories. 
- It helps us to identify the categories with the `highest` and `lowest` number of products in the datset. 

### Sub-Category Analysis

```python
plt.figure(figsize = (8,4))
sns.countplot(data = df, x = 'Sub_Category',
              order = df['Sub_Category'].value_counts().index,palette = 'Set1')
plt.title('Distribution of products by Sub-Category')
plt.xlabel('Number of Products')
plt.ylabel('Sub_Category')
plt.xticks(rotation = 90)
plt.show()
```

## `INSIGHT`
- `Sport Shoes` has the `highest` number of products in the dataset.
- `Sport bras` has the `lowest` number of products.
- The remaining sub-categories have diffrent product count.
- This shows that the products are the not evenly distributed acroess sub-categories.

## Gender wise Analysis 

```python
plt.figure(figsize = (7,5))
sns.countplot(data = df, x = 'Gender',
              order = df['Gender'].value_counts().index,palette={
              'Women': 'pink',
              'Men': 'skyblue'})
plt.title('Distribution of products by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Products')
plt.show()
```

## `INSIGHT`
- The dataset contains products for `Women` and `Men`.
- Women's products are higher in number than Men's products.


## Brand-wise Analysis

```python
brand_count = (
    df.groupby('Brand')
      .size()
      .reset_index(name='Count')
      .sort_values('Count', ascending=False)
)

print(brand_count)
```

```python
plt.figure(figsize=(6,8))

plt.pie(
    brand_count.head(10)['Count'],
    labels=brand_count.head(10)['Brand'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 10 Brands by Number of Products')
plt.show()
```

## `INSIGHT`
- The top 10 brands show a fairly even distribution, so the dataset is not highly concentrated on one brand.
- The presence of several major brands makes the dataset suitable for brand-wise comparison of price, rating and discount.
- Smaller shares of some brands indicate that they have fewer products in the dataset, which may affect brand-level comparisons.
- Overall, the dataset has good brand diversity, allowing us to study differences between brands in further analysis.

```python
brand_avg_rating = brand_avg_rating.sort_values('Rating')

plt.figure(figsize=(8,5))

plt.plot(
    brand_avg_rating['Rating'],
    brand_avg_rating['Brand'],
    '^',
    color='green'
)

plt.title('Top 10 Brands by Average Rating')
plt.xlabel('Average Rating')
plt.ylabel('Brand')

plt.xlim(4.3, 4.7)
plt.show()
```

- `Levis has the lowest average rating` around `4.35`, among the 10 brands.
- Most brands have an `average rating between 4.4 and 4.6`, so the ratings are fairly close.
- The difference between the highest and lowest average ratings is small, showing that all these brands are highly rated.

## Heat map

```python
plt.figure(figsize=(8,6))

corr = df[['MRP', 'Selling_Price', 'Discount_Pct',
           'Rating', 'Rating_Count']].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```
